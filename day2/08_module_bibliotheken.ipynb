## 1 Module importieren

Mit **import modul** bindest du ein Modul ein; Zugriff auf Funktionen ueber **modul.funktion**. Mit **from modul import name** importierst du nur bestimmte Namen; dann kannst du sie ohne modul. verwenden. **from modul import *** importiert alle oeffentlichen Namen (oft unuebersichtlich). Die **Standardbibliothek** gehoert zu Python: math, os, sys, datetime, csv, json und viele andere – keine Installation noetig.


```python
# Beispiel: math und random
import math
print(math.sqrt(16))
print(math.pi)
print(math.ceil(3.2), math.floor(3.8))

import random
print(random.randint(1, 10))
print(random.choice(["a", "b", "c"]))
```

### 1.1 datetime

Das Modul **datetime** bietet **date**, **time**, **datetime** und **timedelta**. **datetime.date.today()** liefert das heutige Datum; **datetime.datetime.now()** Datum und Uhrzeit. Nuetzlich fuer Logs und Zeitstempel.


```python
# Verwendung: datetime
import datetime
heute = datetime.date.today()
print(heute)
jetzt = datetime.datetime.now()
print(jetzt)
print(jetzt.strftime("%Y-%m-%d %H:%M"))
```

## 2 CSV-Dateien mit dem Modul csv

Das Modul **csv** liest und schreibt CSV-Dateien korrekt (Kommas in Anführungszeichen, Zeilenumbrueche). **csv.reader(f)** liefert einen Iterator ueber Zeilen als Listen. Die erste Zeile ist oft der Header. **csv.DictReader(f)** liefert pro Zeile ein Dictionary mit den Header-Namen als Schluessel – sehr praktisch fuer data/sales_data.csv und data/employee_data.csv.


```python
# Beispiel: data/employee_data.csv mit csv.DictReader
import csv
with open("data/employee_data.csv", "r", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    for row in reader:
        print(row["Name"], row["Department"], row["Salary"])
```

### 2.1 Externe Pakete mit pip

```python
# Verwendung: Pruefen, ob ein Paket verfuegbar ist
try:
    import numpy as np
    print("NumPy version:", np.__version__)
except ImportError:
    print("NumPy nicht installiert. z.B.: pip install numpy")
```

```python
# Musterloesung (a)
import datetime
print(datetime.date.today())
from math import pi
print(pi)

# Musterloesung (b)
import csv
with open("data/sales_data.csv", "r", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    for row in reader:
        print(row["Product"], row["Sales"])

# Musterloesung (c)
gehaelter = []
with open("data/employee_data.csv", "r", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    for row in reader:
        gehaelter.append(int(row["Salary"]))
durchschnitt = sum(gehaelter) / len(gehaelter)
print("Durchschnittsgehalt:", durchschnitt)

# Musterloesung (d)
max_umsatz = 0
max_produkt = ""
with open("data/sales_data.csv", "r", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    for row in reader:
        umsatz = int(row["Sales"])
        if umsatz > max_umsatz:
            max_umsatz = umsatz
            max_produkt = row["Product"]
print("Hoechster Umsatz:", max_produkt, max_umsatz)
```
