## 1 Interactive Input and Output

**input(prompt)** reads a line from the console and returns it as a **string**. When you need numbers, you must convert: int(input(...)) or float(input(...)) – invalid input causes a ValueError. **print()** outputs values; multiple arguments are separated by spaces by default. With **sep=** and **end=** you can change the separator and line ending.


```python
# Example: print and format (without interactive input in notebook)
name = "Python"
version = 3.12
print(name, version)
print("{} Version {}".format(name, version))
print(f"{name} Version {version}")
print("Number: {0}, again: {0}".format(42))
```

### 1.1 Number Formatting

In format() or f-strings you can use format specifiers: e.g. **{:.2f}** for two decimal places, **{:d}** for integer. This is useful for tables and reports.


```python
# Usage: Number formatting
x = 3.14159
print("{:.2f}".format(x))
print(f"{x:.3f}")
print("Percent: {:.1%}".format(0.756))
```

## 2 Reading and Writing Files

With **open(path, mode)** you open a file. Mode: **"r"** read (default), **"w"** write (overwrites), **"a"** append, **"r"** text mode. In text mode lines end with \\n. With **with open(...) as f:** the file is automatically closed after the block, even on errors. **f.read()** reads everything; **f.readline()** one line; **f.readlines()** all lines as a list. **f.write(text)** writes (in text mode no automatic newline – write \\n yourself).
