## 1 Funktionen definieren und aufrufen

Mit **def name(parameter):** wird eine Funktion definiert. Der Block darunter (eingerueckt) ist der Funktionskoerper. **return** wert beendet die Funktion und gibt wert zurueck. Ohne return oder mit return ohne Argument gibt die Funktion **None** zurueck. Beim Aufruf name(argumente) werden die Parameter mit den Argumenten gefuellt.


```python
# Beispiel: Einfache Funktion
def begruessung(name):
    return "Hallo, " + name + "!"

print(begruessung("Welt"))
ergebnis = begruessung("Python")
print(ergebnis)
```

### 1.1 Mehrere Parameter und lokale Variablen

Funktionen koennen mehrere Parameter haben. Variablen, die du innerhalb der Funktion zuweist, sind **lokal**: Sie existieren nur waehrend des Aufrufs und sind von aussen nicht sichtbar. Du kannst globale Variablen lesen, aber Zuweisung an einen Namen erzeugt eine lokale Variable, sofern der Name nicht mit **global** deklariert ist (global solltest du sparsam einsetzen).


```python
# Verwendung: Mehrere Parameter
def flaeche(breite, hoehe):
    return breite * hoehe

print(flaeche(4, 5))
print(flaeche(10, 3))
```

## 2 Parameter mit Default-Werten

Parameter koennen **Standardwerte** haben: def f(a, b=10):. Beim Aufruf kannst du b weglassen (dann wird 10 genommen) oder explizit setzen. Parameter mit Default muessen in der Parameterliste **hinter** Parametern ohne Default stehen. Beim Aufruf kannst du Argumente **benannt** uebergeben: f(b=5, a=1) – dann ist die Reihenfolge egal.


```python
# Beispiel: Default-Parameter
def potenz(basis, exponent=2):
    return basis ** exponent

print(potenz(3))
print(potenz(3, 3))
print(potenz(2, exponent=4))
print(potenz(exponent=3, basis=2))
```

### 2.1 Mehrere Rueckgabewerte

Du kannst mit **return a, b** mehrere Werte zurueckgeben. Python packt sie zu einem **Tupel**; der Aufrufer kann sie mit **x, y = f()** entpacken oder als ein Tupel empfangen.


```python
# Verwendung: Mehrere Rueckgabewerte
def min_max(lst):
    if not lst:
        return None, None
    return min(lst), max(lst)

zahlen = [3, 1, 4, 1, 5]
klein, gross = min_max(zahlen)
print(klein, gross)
```
