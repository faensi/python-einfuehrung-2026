## 1 while und for

**while** Bedingung: wiederholt den Block, solange die Bedingung True ist. Du musst die Laufvariable (oder Bedingung) im Block so aendern, dass die Schleife irgendwann endet. **for** element in sequenz: iteriert ueber jedes Element der Sequenz (Liste, String, range, spater auch Datei-Zeilen). **range(start, ende, schritt)** erzeugt Zahlen von start bis vor ende; range(n) ist 0 bis n-1.


```python
# Beispiel: while und for
n = 0
while n < 3:
    print(n)
    n += 1
print("---")
for i in range(3):
    print(i)
for c in "abc":
    print(c)
```

### 1.1 range() genauer

range(stop) liefert 0, 1, ..., stop-1. range(start, stop) von start bis stop-1. range(start, stop, step) mit Schrittweite step (kann negativ sein). range erzeugt keine Liste im Speicher, sondern einen Bereichsobjekt; mit list(range(...)) siehst du die Zahlen.


```python
# Verwendung: range-Varianten
print(list(range(5)))
print(list(range(2, 8)))
print(list(range(0, 10, 2)))
print(list(range(5, 0, -1)))
```

## 2 continue, break, else

**break** beendet die Schleife sofort (nur die innerste Schleife). **continue** springt zum naechsten Durchlauf und ueberspringt den Rest des aktuellen Durchlaufs. Der **else**-Zweig einer for- oder while-Schleife wird genau dann ausgefuehrt, wenn die Schleife **normal** beendet wurde, also nicht durch break. Das ist nuetzlich z.B. bei Suche: "Falls gefunden, break; sonst (else) nichts gefunden".


```python
# Beispiel: break und else
for x in range(5):
    if x == 3:
        break
    print(x)
else:
    print("Schleife normal beendet")
print("---")
for x in range(2):
    print(x)
else:
    print("Schleife normal beendet")
```

### 2.1 Dateien zeilenweise lesen

Eine Datei kannst du in einer for-Schleife direkt zeilenweise iterieren: **for zeile in open(...)** oder besser **for zeile in f:** innerhalb von **with open(...) as f:**. Jede zeile endet mit Zeilenumbruch (\\n); mit zeile.strip() entfernst du ihn. So kannst du data/measurements.log oder data/beispiel_text.txt Zeile fuer Zeile verarbeiten.


```python
# Verwendung: data/measurements.log zeilenweise lesen
with open("data/measurements.log", "r") as f:
    for zeile in f:
        z = zeile.strip()
        if z:
            print(z)
```

```python
# Musterloesung (a)
for i in range(0, 10, 2):
    print(i)

# Musterloesung (b)
s = 0
for i in range(1, 11):
    s += i
print(s)

# Musterloesung (c)
zahlen = [3, 7, 2, 9, 1]
for z in zahlen:
    if z > 5:
        print(z)

# Musterloesung (d)
werte = []
with open("data/measurements.log", "r") as f:
    for zeile in f:
        zeile = zeile.strip()
        if zeile:
            teil = zeile.split(":")
            if len(teil) >= 2:
                wert = float(teil[1].strip())
                werte.append(wert)
print(werte)
if werte:
    mittel = sum(werte) / len(werte)
    print("Mittelwert:", mittel)
```
