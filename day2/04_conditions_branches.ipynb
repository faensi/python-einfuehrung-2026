## 1 Branching with if, elif, else

Branching controls which code is executed. **if** condition: is executed only when the condition is True. **elif** (else if) checks further conditions once the previous ones were False. **else** catches all other cases. At most **one** block from the if/elif/else chain is executed. A block consists of all indented lines under the condition.


```python
# Example: if, elif, else
x = 10
if x < 0:
    print("negative")
elif x == 0:
    print("zero")
else:
    print("positive")
```

### 1.1 Multiple elif Branches

You can add as many elif branches as you like. The conditions are checked from top to bottom; on the first True, the corresponding block is executed and the rest is skipped.


```python
# Usage: Classification by measurement range (e.g. concentration)
konzentration = 0.75
if konzentration >= 0.9:
    stufe = "very high"
elif konzentration >= 0.8:
    stufe = "high"
elif konzentration >= 0.7:
    stufe = "medium"
elif konzentration >= 0.6:
    stufe = "low"
else:
    stufe = "below detection limit"
print(stufe)
```

## 2 Boolean Expressions

Comparisons yield True or False: **==** (equal), **!=** (not equal), **<**, **<=**, **>**, **>=**. Combination with **and**, **or**, **not**. Other useful operators: **in** (element in sequence), **is** (identity, e.g. comparison with None). With **and** and **or** evaluation is left to right; as soon as the result is determined, evaluation stops (short-circuit).


```python
print(False and False)
print(True and False)
print(True and True)
print(False and True)
print(not(False or False))
print(True or False)
print(True or True)
print(False or True)
print(False ^ False) # XOR
```

```python
# Example: and, or, in
temperatur = 22.5
print(temperatur >= 18 and temperatur <= 26)
name = "Python"
print("P" in name)
print(not (temperatur < 0))
print(True or False)
print(False and True)

```

### 2.1 Nested if Statements

Within an if block you can write further if/elif/else. Pay attention to clear indentation: each block belongs to the immediately enclosing if/elif/else.


```python
# Usage: Nested
x, y = 5, 10
if x > 0:
    if y > 0:
        print("Both positive")
    else:
        print("x positive, y not")
else:
    print("x not positive")
```
