## 1 while and for

**while** condition: repeats the block as long as the condition is True. You must change the loop variable (or condition) in the block so that the loop eventually ends. **for** element in sequence: iterates over each element of the sequence (list, string, range, later also file lines). **range(start, stop, step)** generates numbers from start up to but not including stop; range(n) is 0 to n-1.


```python
# Example: while and for
n = 0
while n < 3:
    print(n)
    n += 1
print("---")
for i in range(3):
    print(i)
for c in "abc":
    print(c)
```

### 1.1 range() in Detail

range(stop) yields 0, 1, ..., stop-1. range(start, stop) from start to stop-1. range(start, stop, step) with step size (can be negative). range does not create a list in memory, but a range object; with list(range(...)) you see the numbers.


```python
# Usage: range variants
print(list(range(5)))
print(list(range(2, 8)))
print(list(range(0, 10, 2)))
print(list(range(5, 0, -1)))
```

## 2 continue, break, else

**break** terminates the loop immediately (only the innermost loop). **continue** jumps to the next iteration and skips the rest of the current iteration. The **else** branch of a for or while loop is executed exactly when the loop terminated **normally**, i.e. not by break. This is useful e.g. for search: "If found, break; otherwise (else) nothing found".


```python
# Example: break and else
for x in range(5):
    if x == 3:
        break
    print(x)
else:
    print("Loop ended normally")
print("---")
for x in range(2):
    print(x)
else:
    print("Loop ended normally")
```

### 2.1 Reading Files Line by Line

You can iterate over a file line by line in a for loop: **for line in open(...)** or better **for line in f:** within **with open(...) as f:**. Each line ends with a newline (\\n); with line.strip() you remove it. This way you can process data/measurements.log or data/example_text.txt line by line.


```python
# Usage: read data/measurements.log line by line
with open("data/measurements.log", "r") as f:
    for line in f:
        z = line.strip()
        if z:
            print(z)
```
