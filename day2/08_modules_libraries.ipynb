## 1 Importing Modules

With **import module** you include a module; access functions via **module.function**. With **from module import name** you import only specific names; then you can use them without the module prefix. **from module import *** imports all public names (often unclear). The **standard library** is part of Python: math, os, sys, datetime, csv, json and many others – no installation needed.


```python
# Example: math and random
import math
print(math.sqrt(16))
print(math.pi)
print(math.ceil(3.2), math.floor(3.8))

import random
print(random.randint(1, 10))
print(random.choice(["a", "b", "c"]))
```

### 1.1 datetime

The **datetime** module provides **date**, **time**, **datetime** and **timedelta**. **datetime.date.today()** returns today's date; **datetime.datetime.now()** date and time. Useful for logs and timestamps.


```python
# Usage: datetime
import datetime
heute = datetime.date.today()
print(heute)
jetzt = datetime.datetime.now()
print(jetzt)
print(jetzt.strftime("%Y-%m-%d %H:%M"))
```

## 2 CSV Files with the csv Module

The **csv** module reads and writes CSV files correctly (commas in quotes, line breaks). **csv.reader(f)** yields an iterator over lines as lists. The first line is often the header. **csv.DictReader(f)** yields a dictionary per line with the header names as keys – very practical for data/sales_data.csv and data/employee_data.csv.


```python
# Example: data/employee_data.csv with csv.DictReader
import csv
with open("data/employee_data.csv", "r", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    for row in reader:
        print(row["Name"], row["Department"], row["Salary"])
```

### 2.1 External Packages with pip

Packages that are not part of the standard library are installed with **pip install package_name** (or in Conda: **conda install package_name**). The **PyPI** registry (pypi.org) lists thousands of packages. After installation you can import them like standard modules (e.g. import numpy).


```python
# Usage: Check if a package is available
try:
    import numpy as np
    print("NumPy version:", np.__version__)
except ImportError:
    print("NumPy not installed. E.g.: pip install numpy")
```

#### Solution:


```python
# Model solution (a)
import datetime
print(datetime.date.today())
from math import pi
print(pi)

# Model solution (b)
import csv
with open("data/sales_data.csv", "r", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    for row in reader:
        print(row["Product"], row["Sales"])

# Model solution (c)
gehaelter = []
with open("data/employee_data.csv", "r", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    for row in reader:
        gehaelter.append(int(row["Salary"]))
durchschnitt = sum(gehaelter) / len(gehaelter)
print("Average salary:", durchschnitt)

# Model solution (d)
max_umsatz = 0
max_produkt = ""
with open("data/sales_data.csv", "r", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    for row in reader:
        umsatz = int(row["Sales"])
        if umsatz > max_umsatz:
            max_umsatz = umsatz
            max_produkt = row["Product"]
print("Highest sales:", max_produkt, max_umsatz)

# Model solution (e)
import random
import csv
with open("data/sales_data.csv", "r", encoding="utf-8") as f:
    rows = list(csv.DictReader(f))
picked = random.choice(rows)
print("Random pick:", picked["Product"], picked["Sales"])

# Model solution (f)
max_salary = 0
with open("data/employee_data.csv", "r", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    for row in reader:
        sal = int(row["Salary"])
        if sal > max_salary:
            max_salary = sal

top_names = []
with open("data/employee_data.csv", "r", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    for row in reader:
        if int(row["Salary"]) == max_salary:
            top_names.append(row["Name"])
print("Highest salary:", max_salary, "Employees:", top_names)
```
