## 1 Verzweigungen mit if, elif, else

Verzweigungen steuern, welcher Code ausgefuehrt wird. **if** Bedingung: wird nur ausgefuehrt, wenn die Bedingung True ist. **elif** (else if) prueft weitere Bedingungen, sobald die vorherigen False waren. **else** faengt alle uebrigen Faelle. Es wird hoechstens **ein** Block aus der if/elif/else-Kette ausgefuehrt. Ein Block besteht aus allen eingerueckten Zeilen unter der Bedingung.


```python
if CONDITION :
    STATEMENT
elif CONDITION:
    STATEMENT
else:
    STATEMENT


if CONDITION:
    STATEMENT
else:
    if CONDITION:
        STATEMENT
    else:
        STATEMENT
```

```python
# Beispiel: if, elif, else
x = 10
if x < 0:
    print("negativ")
elif x == 0:
    print("null")
else:
    print("positiv")
```

### 1.1 Mehrere elif-Zweige

Du kannst beliebig viele elif-Zweige anhaengen. Die Bedingungen werden von oben nach unten geprueft; beim ersten True wird der zugehoerige Block ausgefuehrt und der Rest uebersprungen.


```python
# Verwendung: Einstufung nach Messbereich (z.B. Konzentration)
konzentration = 0.75
if konzentration >= 0.9:
    stufe = "sehr hoch"
elif konzentration >= 0.8:
    stufe = "hoch"
elif konzentration >= 0.7:
    stufe = "mittel"
elif konzentration >= 0.6:
    stufe = "niedrig"
else:
    stufe = "unter Nachweisgrenze"
print(stufe)
```

## 2 Boolsche Ausdruecke

Vergleiche liefern True oder False: **==** (gleich), **!=** (ungleich), **<**, **<=**, **>**, **>=**. Verknuepfung mit **and**, **or**, **not**. Weitere nuetzliche Operatoren: **in** (Element in Sequenz), **is** (Identitaet, z.B. Vergleich mit None). Bei **and** und **or** wird von links nach rechts ausgewertet; sobald das Ergebnis feststeht, wird abgebrochen (Kurzschluss).


```python
print(False and False)
print(True and False)
print(True and True)
print(False and True)
```

```python
print(not(False or False))
print(True or False)
print(True or True)
print(False or True)

print(False ^ False) # XOR
```

```python
# Beispiel: and, or, in
temperatur = 22.5
print(temperatur >= 18 and temperatur <= 26)
name = "Python"
print("P" in name)
print(not (temperatur < 0))
print(True or False)
print(False and True)

```

### 2.1 Verschachtelte if-Statements

Innerhalb eines if-Blocks kannst du weitere if/elif/else schreiben. Achte auf klare Einrueckung: Jeder Block gehoert zu dem naechsthoeheren if/elif/else.


```python
# Verwendung: Verschachtelt
x, y = 5, 10
if x > 0:
    if y > 0:
        print("Beide positiv")
    else:
        print("x positiv, y nicht")
else:
    print("x nicht positiv")
```
