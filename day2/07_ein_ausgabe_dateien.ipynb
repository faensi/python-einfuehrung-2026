## 1 Interaktive Eingabe und Ausgabe

**input(prompt)** liest eine Zeile von der Konsole und gibt sie als **String** zurueck. Wenn du Zahlen brauchst, musst du umwandeln: int(input(...)) oder float(input(...)) – bei ungueltiger Eingabe entsteht ein ValueError. **print()** gibt Werte aus; mehrere Argumente werden standardmaessig durch Leerzeichen getrennt. Mit **sep=** und **end=** kannst du Trennzeichen und Zeilenende aendern.


```python
# Beispiel: print und format (ohne interaktive Eingabe im Notebook)
name = "Python"
version = 3.12
print(name, version)
print("{} Version {}".format(name, version))
print(f"{name} Version {version}")
print("Zahl: {0}, nochmal: {0}".format(42))
```

### 1.1 Formatierung von Zahlen

In format() oder f-Strings kannst du Formatangaben verwenden: z.B. **{:.2f}** fuer zwei Nachkommastellen, **{:d}** fuer Integer. Das ist nuetzlich fuer Tabellen und Berichte.


```python
# Verwendung: Zahlen formatieren
x = 3.14159
print("{:.2f}".format(x))
print(f"{x:.3f}")
print("Prozent: {:.1%}".format(0.756))
```

## 2 Dateien lesen und schreiben

Mit **open(pfad, modus)** oeffnest du eine Datei. Modus: **"r"** lesen (Standard), **"w"** schreiben (ueberschreibt), **"a"** anhaengen, **"r"** Textmodus. Im Textmodus werden Zeilen mit \\n beendet. Mit **with open(...) as f:** wird die Datei nach dem Block automatisch geschlossen, auch bei Fehlern. **f.read()** liest alles; **f.readline()** eine Zeile; **f.readlines()** alle Zeilen als Liste. **f.write(text)** schreibt (bei Textmodus keine automatische Zeilenschaltung – \\n selbst schreiben).


```python
# Beispiel: Datei schreiben und lesen
with open("data/beispiel_ausgabe.txt", "w") as f:
    f.write("Zeile 1\n")
    f.write("Zeile 2\n")

with open("data/beispiel_ausgabe.txt", "r") as f:
    inhalt = f.read()
print(repr(inhalt))

with open("data/beispiel_ausgabe.txt", "r") as f:
    for zeile in f:
        print(zeile.strip())
```

### 2.1 Vorhandene Daten in data/

Im Ordner **data/** liegen u.a.: **beispiel_text.txt** (mehrere Zeilen Text), **measurements.log** (Messung N: Wert), **errors.log** (Zeilen mit WARNING/ERROR), **sales_data.csv** und **employee_data.csv** (CSV mit Header). Du kannst diese Dateien mit open() zeilenweise lesen und auswerten. Fuer CSV mit Komma-Trennung: erste Zeile ist Header, dann zeile.split(",") fuer jede Datenzeile.


```python
# Verwendung: data/errors.log lesen und Zeilen mit ERROR zaehlen
with open("data/errors.log", "r") as f:
    anzahl_error = 0
    for zeile in f:
        if "ERROR" in zeile:
            anzahl_error += 1
print("Anzahl Zeilen mit ERROR:", anzahl_error)
```

```python
# Musterloesung (a)
with open("data/zahlen.txt", "w") as f:
    for i in [1, 2, 3]:
        f.write(str(i) + "\n")

with open("data/zahlen.txt", "r") as f:
    zeilen = f.readlines()
summe = 0
for z in zeilen:
    summe += int(z.strip())
print(summe)

# Musterloesung (b)
werte = []
with open("data/measurements.log", "r") as f:
    for zeile in f:
        zeile = zeile.strip()
        if ":" in zeile:
            wert = float(zeile.split(":")[1].strip())
            werte.append(wert)
mittel = sum(werte) / len(werte) if werte else 0
with open("data/ergebnis.txt", "w") as f:
    f.write("Mittelwert: {:.2f}\n".format(mittel))
print("Mittelwert:", mittel)

# Musterloesung (c)
warn, err = 0, 0
with open("data/errors.log", "r") as f:
    for zeile in f:
        if "WARNING" in zeile:
            warn += 1
        if "ERROR" in zeile:
            err += 1
print("WARNING:", warn, "ERROR:", err)
with open("data/fehler_zusammenfassung.txt", "w") as f:
    f.write("WARNING: {}\n".format(warn))
    f.write("ERROR: {}\n".format(err))

# Musterloesung (d)
with open("data/sales_data.csv", "r") as f:
    header = f.readline()
    erste_datenzeile = f.readline().strip()
felder = erste_datenzeile.split(",")
produkt, kategorie, umsatz = felder[0], felder[1], felder[2]
print("Produkt:", produkt, "Kategorie:", kategorie, "Umsatz:", umsatz)
```
