## 1 Defining and Calling Functions

With **def name(parameters):** a function is defined. The block below (indented) is the function body. **return** value terminates the function and returns value. Without return or with return without argument, the function returns **None**. When calling name(arguments), the parameters are filled with the arguments.


```python
# Example: Simple function
def greeting(name):
    return "Hello, " + name + "!"

print(greeting("World"))
result = greeting("Python")
print(result)
```

### 1.1 Multiple Parameters and Local Variables

Functions can have multiple parameters. Variables you assign within the function are **local**: They exist only during the call and are not visible from outside. You can read global variables, but assignment to a name creates a local variable unless the name is declared with **global** (you should use global sparingly).


```python
# Usage: Multiple parameters
def area(width, height):
    return width * height

print(area(4, 5))
print(area(10, 3))
```

## 2 Parameters with Default Values

Parameters can have **default values**: def f(a, b=10):. When calling, you can omit b (then 10 is used) or set it explicitly. Parameters with defaults must come **after** parameters without defaults in the parameter list. When calling you can pass arguments **by name**: f(b=5, a=1) – then the order does not matter.


```python
# Example: Default parameters
def power(base, exponent=2):
    return base ** exponent

print(power(3))
print(power(3, 3))
print(power(2, exponent=4))
print(power(exponent=3, base=2))
```

### 2.1 Multiple Return Values

You can return multiple values with **return a, b**. Python packs them into a **tuple**; the caller can unpack them with **x, y = f()** or receive them as a tuple.


```python
# Usage: Multiple return values
def min_max(lst):
    if not lst:
        return None, None
    return min(lst), max(lst)

numbers = [3, 1, 4, 1, 5]
small, large = min_max(numbers)
print(small, large)
```
