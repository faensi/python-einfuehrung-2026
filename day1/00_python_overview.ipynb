## 1 What is Python?

Python is an **interpreted**, **dynamically typed** programming language. This means: you write source code, and a program (the interpreter) executes it directly, without having to compile first. The **type** of a variable (e.g. whether it is a number or text) is determined at runtime, not in the source code declaration.

Python places great emphasis on **readability**: clear syntax, indentation as a structural element, and a philosophy document ("The Zen of Python") that formulates guidelines for good Python code.


```python
# Usage: First steps - running Python
print("Hello, Python!")
print(2 + 3)
print("Python version:", __import__("sys").version.split()[0])  # __import__ is a dunder function that allows importing modules dynamically

# import sys
# print(sys.version.split())
```

## 2 Differences from other languages

If you know C, C++, Java or similar languages, you will notice:

- **No curly braces**: Blocks (e.g. after if, for, def) are defined by **indentation** (spaces or tabs). Same indentation level = same block.
- **No explicit type for variables**: You do not write int x = 5; but x = 5. The type of x is automatically recognised as int.
- **Interpreted**: There is no separate compile step; the interpreter reads and executes the code (optionally there is also bytecode and .pyc files for acceleration).
- **Extensive standard library**: Many tasks (files, network, mathematics, dates) can be solved without additional installation. In addition, there is the **PyPI** ecosystem and the **pip** tool for thousands of packages.


```python
# Example: Indentation defines the block
x = 10
if x > 5:
    print("x is greater than 5")
    print("This still belongs to the if block")
print("This is outside the if")
```

### 2.1 Dynamic typing

A variable can take values of different types one after another. This is often not allowed in statically typed languages or only possible with special constructs.


```python
# Usage: Same variable, different types
z = 42
print(type(z), z)
z = "forty-two"
print(type(z), z)
z = [1, 2, 3]
print(type(z), z)
```

## 3 Output with print

The **print()** function outputs values to the console. You can pass multiple arguments; they are separated by spaces by default. The newline at the end can be changed with the **end** parameter.


```python
# Example: print with multiple arguments and end
print("One", "Two", "Three")
print("Line 1", end=" ")
print("Line 2 (same output line)")
print("Separators:", "A", "B", "C", sep=" | ")
```
