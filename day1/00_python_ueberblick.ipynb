## 1 Was ist Python?

Python ist eine **interpretierte**, **dynamisch typisierte** Programmiersprache. Das bedeutet: Du schreibst Quellcode, und ein Programm (der Interpreter) fuehrt ihn direkt aus, ohne dass du zuerst kompilieren musst. Der **Typ** einer Variable (z.B. ob es eine Zahl oder ein Text ist) wird zur Laufzeit bestimmt, nicht in der Quellcode-Deklaration.

Python legt grossen Wert auf **Lesbarkeit**: klare Syntax, Einrueckung als Strukturelement, und ein Philosophie-Dokument ("The Zen of Python"), das Richtlinien fuer guten Python-Code formuliert.


```python
# Verwendung: Erste Schritte - Python ausfuehren
print("Hallo, Python!")
print(2 + 3)
print("Python-Version:", __import__("sys").version.split()[0]) # __import__ ist eine dunder-Function, die es ermöglicht, Module dynamisch zu importieren

# import sys
# print(sys.version.split())
```

## 2 Unterschiede zu anderen Sprachen

Wenn du C, C++, Java oder aehnliche Sprachen kennst, fallen dir folgende Punkte auf:

- **Keine geschweiften Klammern**: Bloecke (z.B. nach if, for, def) werden durch **Einrueckung** (Leerzeichen oder Tabs) definiert. Gleiche Einrueckungsebene = gleicher Block.
- **Kein expliziter Typ bei Variablen**: Du schreibst nicht int x = 5;, sondern x = 5. Der Typ von x wird automatisch als int erkannt.
- **Interpretiert**: Es gibt keinen separaten Kompilier-Schritt; der Interpreter liest und fuehrt den Code aus (optional gibt es auch Bytecode und .pyc-Dateien zur Beschleunigung).
- **Umfangreiche Standardbibliothek**: Viele Aufgaben (Dateien, Netzwerk, Mathematik, Datum) sind ohne zusaetzliche Installation loesbar. Zusaetzlich gibt es das Oekosystem **PyPI** und das Tool **pip** fuer tausende Pakete.


```python
# Beispiel: Einrueckung definiert den Block
x = 10
if x > 5:
    print("x ist groesser als 5")
    print("Dies gehoert noch zum if-Block")
print("Dies steht ausserhalb des if")
```

### 2.1 Dynamische Typisierung

Eine Variable kann nacheinander Werte unterschiedlichen Typs annehmen. Das ist in statisch typisierten Sprachen oft nicht erlaubt oder nur mit besonderen Konstrukten moeglich.


```python
# Verwendung: Dieselbe Variable, verschiedene Typen
z = 42
print(type(z), z)
z = "zweiundvierzig"
print(type(z), z)
z = [1, 2, 3]
print(type(z), z)
```

## 3 Ausgabe mit print

Die Funktion **print()** gibt Werte auf die Konsole aus. Du kannst mehrere Argumente uebergeben; sie werden standardmaessig durch Leerzeichen getrennt. Ein Zeilenumbruch am Ende kann mit dem Parameter **end** geaendert werden.


```python
# Beispiel: print mit mehreren Argumenten und end
print("Eins", "Zwei", "Drei")
print("Zeile 1", end=" ")
print("Zeile 2 (gleiche Ausgabezeile)")
print("Separatoren:", "A", "B", "C", sep=" | ")
```
