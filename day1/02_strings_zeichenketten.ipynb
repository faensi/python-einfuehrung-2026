## 1 Strings erzeugen und Indizierung

Strings sind **Zeichenketten** in einfachen oder doppelten Anführungszeichen. Sie sind **unveraenderlich** (immutable): Eine Operation wie s.upper() erzeugt einen neuen String, aendert s selbst nicht. Der Zugriff auf einzelne Zeichen erfolgt ueber den **Index** (0-basiert). Negative Indizes zaehlen vom Ende: -1 ist das letzte Zeichen.


```python
# Beispiel: Indizierung
s = "Python"
print(s[0], s[1], s[-1], s[-2])
print(len(s))
# s[0] = "p"  # Fehler: Strings sind immutable
```

### 1.1 Slicing

Mit der **Slice-Notation** s[start:ende:schritt] holst du einen Teilstring. start ist inklusive, ende exklusive. Fehlt start, wird vom Anfang genommen; fehlt ende, bis zum Ende. Ein negativer schritt kehrt die Reihenfolge um (z.B. s[::-1] fuer den String rueckwaerts).


```python
# Verwendung: Slicing
s = "Python"
print(s[1:4])
print(s[:3], s[3:])
print(s[::2])
print(s[::-1])
print(s[-3:])
```

## 2 Wichtige String-Methoden

Strings haben viele **Methoden**: z.B. **upper()**, **lower()** fuer Gross-/Kleinschreibung, **strip()** zum Entfernen von Leerzeichen am Anfang und Ende, **split(trennzeichen)** zum Aufteilen in eine Liste von Teilstrings, **join(liste)** zum Zusammenfuegen, **replace(alt, neu)**, **format()** oder f-Strings fuer formatierte Ausgabe, **find()**, **startswith()**, **endswith()**, **in** zum Pruefen von Teilstrings.


```python
# Beispiel: split, join, replace
text = "  Hallo, Welt!  "
print(repr(text.strip()))
woerter = "eins zwei drei".split()
print(woerter)
print("-".join(woerter))
print("Hallo Welt".replace("Welt", "Python"))
```

### 2.1 Formatierte Ausgabe

Mit der **format()**-Methode oder **f-Strings** (vor Python 3.6: f"...{variable}...") kannst du Werte in einen String einbauen. Das ist besonders fuer Ausgaben und Logs nuetzlich.


```python
# Verwendung: format und f-String
name, alter = "Anna", 30
print("Name: {} Alter: {}".format(name, alter))
print(f"Name: {name} Alter: {alter}")
print("Zahl: {} und Text: {}".format(42, "Python"))
```

## 3 Text aus Dateien lesen (data/)

Mit **open(pfad, "r")** oeffnest du eine Datei zum Lesen. Der Pfad **data/beispiel_text.txt** bezieht sich auf die Datei im Ordner data/ (relativ zum Verzeichnis, aus dem du das Notebook startest). Mit **read()** liest du den gesamten Inhalt als einen String; mit **readlines()** zeilenweise als Liste von Strings. Nach dem Lesen die Datei schliessen oder **with open(...) as f:** verwenden.


```python
# Beispiel: Datei aus data/ lesen
with open("data/beispiel_text.txt", "r") as f:
    inhalt = f.read()
print(inhalt)
print("--- Zeilen als Liste ---")
with open("data/beispiel_text.txt", "r") as f:
    zeilen = f.readlines()
for i, zeile in enumerate(zeilen):
    print(i, repr(zeile.strip()))
```
