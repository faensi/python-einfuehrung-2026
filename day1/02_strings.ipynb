## 1 Creating and Indexing Strings

Strings are **character sequences** in single or double quotes. They are **immutable**: an operation like s.upper() creates a new string, it does not modify s itself. Access to individual characters is via **index** (0-based). Negative indices count from the end: -1 is the last character.


```python
# Example: Indexing
s = "Python"
print(s[0], s[1], s[-1], s[-2])
print(len(s))
# s[0] = "p"  # Error: strings are immutable
```

### 1.1 Slicing

With the **slice notation** s[start:end:step] you extract a substring. start is inclusive, end is exclusive. If start is omitted, it starts from the beginning; if end is omitted, it goes to the end. A negative step reverses the order (e.g. s[::-1] for the string backwards).


```python
# Usage: Slicing
s = "Python"
print(s[1:4])
print(s[:3], s[3:])
print(s[::2])
print(s[::-1])
print(s[-3:])
```

## 2 Important String Methods

Strings have many **methods**: e.g. **upper()**, **lower()** for case conversion, **strip()** to remove leading/trailing whitespace, **split(separator)** to split into a list of substrings, **join(list)** to combine, **replace(old, new)**, **format()** or f-strings for formatted output, **find()**, **startswith()**, **endswith()**, **in** for checking substrings.


```python
# Example: split, join, replace
text = "  Hello, World!  "
print(repr(text.strip()))
words = "one two three".split()
print(words)
print("-".join(words))
print("Hello World".replace("World", "Python"))
```

### 2.1 Formatted Output

With the **format()** method or **f-strings** (since Python 3.6: f"...{variable}...") you can embed values in a string. This is especially useful for output and logs.


```python
# Usage: format and f-string
name, age = "Anna", 30
print("Name: {} Age: {}".format(name, age))
print(f"Name: {name} Age: {age}")
print("Number: {} and text: {}".format(42, "Python"))
```

## 3 Reading Text from Files (data/)

With **open(path, "r")** you open a file for reading. The path **data/example_text.txt** refers to the file in the data/ folder (relative to the directory from which you start the notebook). With **read()** you read the entire content as one string; with **readlines()** line by line as a list of strings. Close the file after reading or use **with open(...) as f:**.


```python
# Example: Read file from data/
with open("data/example_text.txt", "r") as f:
    content = f.read()
print(content)
print("--- Lines as list ---")
with open("data/example_text.txt", "r") as f:
    lines = f.readlines()
for i, line in enumerate(lines):
    print(i, repr(line.strip()))
```
