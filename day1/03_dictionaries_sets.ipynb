## 1 Dictionaries

Ein **Dictionary** (dict) speichert **Schluessel-Wert-Paare**. Der Schluessel muss **hashbar** (unveraenderlich) und innerhalb eines Dicts **eindeutig** sein. Zugriff ueber **d[schluessel]** oder **d.get(schluessel, default)**. Mit get() vermeidest du einen KeyError, wenn der Schluessel fehlt. Du kannst Werte aendern und neue Schluessel hinzufuegen.


```python
# Beispiel: Dictionary anlegen und nutzen
person = {"name": "Anna", "alter": 30, "stadt": "Berlin"}
print(person["name"])
print(person.get("land", "unbekannt"))
person["alter"] = 31
person["beruf"] = "Ingenieurin"
print(person)
print(person.keys())
print(person.values())
```

### 1.1 Iteration ueber Dictionaries

Mit **for k in d:** iterierst du ueber die Schluessel. **for k, v in d.items():** liefert Schluessel und Wert. **d.items()** gibt Paare (key, value) zurueck.


```python
# Verwendung: Iteration
person = {"name": "Anna", "alter": 30}
for key in person:
    print(key, person[key])
for k, v in person.items():
    print(k, "->", v)
```

## 2 Sets und Frozensets

Ein **Set** ist eine **Menge** von eindeutigen Elementen: keine Duplikate, keine feste Reihenfolge. Elemente muessen hashbar sein. Typische Operationen: **add**, **remove**, **in**, **len**, **Vereinigung** (| oder union), **Schnitt** (& oder intersection), **Differenz** (-). Ein **Frozenset** ist ein unveraenderliches Set (z.B. als Dictionary-Schluessel nutzbar).


```python
# Beispiel: Set und Frozenset
zahlen = {1, 2, 3, 2, 1}
print(zahlen)
zahlen.add(4)
zahlen.discard(2)
print(zahlen)
print(1 in zahlen)
fs = frozenset([1, 2, 3])
print(fs)
```

### 2.1 Mengenoperationen

Mit |, &, - und ^ (symmetrische Differenz) kannst du Mengen verknuepfen. Das ist nuetzlich, um z.B. eindeutige Werte aus einer Liste zu gewinnen: set(liste).


```python
# Verwendung: Mengenoperationen
a = {1, 2, 3}
b = {2, 3, 4}
print("Vereinigung:", a | b)
print("Schnitt:", a & b)
print("Differenz a-b:", a - b)
lst = [1, 2, 2, 3, 3, 3]
print("Eindeutige Werte:", set(lst))
```

## 3 Datenstrukturen aus Zeilen aufbauen

Haeufig liest man Zeilen aus einer Datei (z.B. CSV) und baut daraus Listen oder Dictionaries. Beispiel: Eine Zeile "Name;Alter" in **data/beispiel_text.txt** oder eine einfache CSV-Struktur. Mit **split()** zerlegst du Zeilen in Felder; mit einem Dictionary kannst du Werte nach Spaltenname zuordnen. Die echten CSV-Dateien (z.B. data/sales_data.csv, data/employee_data.csv) werden spaeter mit dem Modul csv oder mit Pandas eingelesen; hier siehst du das Prinzip mit einer einfachen Zeile.


```python
# Beispiel: Zeile in Dictionary umwandeln (Header + eine Zeile)
header = "Name,Abteilung,Gehalt"
zeile = "Max,IT,50000"
keys = header.split(",")
values = zeile.split(",")
record = dict(zip(keys, values))
print(record)
print(record["Name"], record["Gehalt"])
```
