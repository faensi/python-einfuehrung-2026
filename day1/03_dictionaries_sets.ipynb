## 1 Dictionaries

A **Dictionary** (dict) stores **key-value pairs**. The key must be **hashable** (immutable) and **unique** within a dict. Access via **d[key]** or **d.get(key, default)**. With get() you avoid a KeyError when the key is missing. You can change values and add new keys.


```python
# Example: Create and use a dictionary
person = {"name": "Anna", "age": 30, "city": "Berlin"}
print(person["name"])
print(person.get("land", "unknown"))
person["age"] = 31
person["profession"] = "engineer"
print(person)
print(person.keys())
print(person.values())
```

### 1.1 Iterating over Dictionaries

With **for k in d:** you iterate over the keys. **for k, v in d.items():** yields key and value. **d.items()** returns pairs (key, value).


```python
# Usage: Iteration
person = {"name": "Anna", "age": 30}
for key in person:
    print(key, person[key])
for k, v in person.items():
    print(k, "->", v)
```

## 2 Sets and Frozensets

A **Set** is a **collection** of unique elements: no duplicates, no fixed order. Elements must be hashable. Typical operations: **add**, **remove**, **in**, **len**, **union** (| or union), **intersection** (& or intersection), **difference** (-). A **Frozenset** is an immutable set (e.g. usable as a dictionary key).


```python
# Example: Set and Frozenset
zahlen = {1, 2, 3, 2, 1}
print(zahlen)
zahlen.add(4)
zahlen.discard(2)
print(zahlen)
print(1 in zahlen)
fs = frozenset([1, 2, 3])
print(fs)
```

### 2.1 Set Operations

With |, &, - and ^ (symmetric difference) you can combine sets. This is useful e.g. to get unique values from a list: set(list).


```python
# Usage: Set operations
a = {1, 2, 3}
b = {2, 3, 4}
print("Union:", a | b)
print("Intersection:", a & b)
print("Difference a-b:", a - b)
lst = [1, 2, 2, 3, 3, 3]
print("Unique values:", set(lst))
```
