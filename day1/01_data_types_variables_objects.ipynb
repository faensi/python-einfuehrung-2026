## 1 Built-in Data Types

Python includes several **built-in data types**. The most important for getting started:
- **int**: Integers (arbitrarily large in Python 3)
- **float**: Floating-point numbers
- **bool**: Boolean values True and False
- **NoneType**: The value None represents "nothing" or "no value" (e.g. as return value of a function without return)


```python
# Example: Types and type()
a = 42
b = 3.14
c = True
d = None
print(type(a), type(b), type(c), type(d))
print(a, b, c, d)
```

### 1.1 Arithmetic with int and float

When mixing int and float in arithmetic, the result is a float. The division **/** in Python 3 always returns a float (even 4 / 2 gives 2.0). Use **//** for integer division and **%** for the remainder.


```python
# Usage: Division and integer operations
print(7 / 2)
print(7 // 2)
print(7 % 2)
print(type(10 / 5))
print(2 ** 10)
```

## 2 Variables and Objects

Variables are **names** that reference **objects** in memory. In Python (almost) everything is an object: numbers, strings, lists, functions. Each object has a **class** (a type). You can determine the type with **type(obj)** or **obj.__class__**. Use **isinstance(obj, Type)** to check whether an object belongs to a type.


```python
# Example: Objects have methods and attributes
x = 5
print(x.__class__)
s = "hello"
print(s.upper())
print(s.__class__)
print(isinstance(s, str))
print(isinstance(x, float))
```

### 2.1 Multiple assignment and swapping

You can assign multiple variables in one line (a, b = 1, 2) and also swap two values (a, b = b, a) without needing a temporary variable.


```python
# Usage: Multiple assignment
name, age = "Anna", 28
print(name, age)
x, y = 10, 20
x, y = y, x
print(x, y)
```

## 3 Lists as Ordered Collections

A **list** is an ordered, mutable sequence of objects. It is created with square brackets: [1, 2, 3]. You can access elements by index (0-based), append with .append(), get the length with len(), and iterate with for. Lists play a central role in almost every Python program.


```python
# Example: Create and use a list
numbers = [10, 20, 30, 40, 50]
print(numbers[0], numbers[-1])
numbers.append(60)
print(len(numbers))
for n in numbers:
    print(n, end=" ")
print()
```
