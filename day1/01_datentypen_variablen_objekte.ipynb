## 1 Eingebaute Datentypen

Python bringt mehrere **eingebaute Datentypen** mit. Die wichtigsten fuer den Einstieg:
- **int**: Ganze Zahlen (beliebig gross in Python 3)
- **float**: Gleitkommazahlen
- **bool**: Wahrheitswerte True und False
- **NoneType**: Der Wert None steht fuer "nichts" oder "kein Wert" (z.B. als Rueckgabewert einer Funktion ohne return)


```python
# Beispiel: Typen und type()
a = 42
b = 3.14
c = True
d = None
print(type(a), type(b), type(c), type(d))
print(a, b, c, d)
```

### 1.1 Rechnen mit int und float

Bei Rechenoperationen zwischen int und float wird das Ergebnis ein float. Die Division **/** liefert in Python 3 immer ein float (auch 4 / 2 ergibt 2.0). Fuer ganzzahlige Division nutzt man **//**, fuer den Rest **%**.


```python
# Verwendung: Division und ganzzahlige Operationen
print(7 / 2)
print(7 // 2)
print(7 % 2)
print(type(10 / 5))
print(2 ** 10)
```

## 2 Variablen und Objekte

Variablen sind **Namen**, die auf **Objekte** im Speicher verweisen. In Python ist (fast) alles ein Objekt: Zahlen, Strings, Listen, Funktionen. Jedes Objekt hat eine **Klasse** (einen Typ). Du kannst mit **type(obj)** oder **obj.__class__** den Typ ermitteln. Mit **isinstance(obj, Typ)** pruefst du, ob ein Objekt zu einem Typ gehoert.


```python
# Beispiel: Objekte haben Methoden und Attribute
x = 5
print(x.__class__)
s = "hallo"
print(s.upper())
print(s.__class__)
print(isinstance(s, str))
print(isinstance(x, float))
```

### 2.1 Mehrfachzuweisung und Tausch

Du kannst mehrere Variablen in einer Zeile zuweisen (a, b = 1, 2) und so auch zwei Werte tauschen (a, b = b, a), ohne eine Hilfsvariable zu brauchen.


```python
# Verwendung: Mehrfachzuweisung
name, alter = "Anna", 28
print(name, alter)
x, y = 10, 20
x, y = y, x
print(x, y)
```

## 3 Listen als geordnete Sammlung

Eine **Liste** (list) ist eine geordnete, veraenderliche Folge von Objekten. Sie wird mit eckigen Klammern erstellt: [1, 2, 3]. Du kannst Elemente ueber den Index ansprechen (0-basiert), anhaengen (.append), die Laenge mit len() abfragen und mit for durchiterieren. Listen spielen in fast jedem Python-Programm eine zentrale Rolle.


```python
# Beispiel: Liste erstellen und nutzen
zahlen = [10, 20, 30, 40, 50]
print(zahlen[0], zahlen[-1])
zahlen.append(60)
print(len(zahlen))
for z in zahlen:
    print(z, end=" ")
print()
```
