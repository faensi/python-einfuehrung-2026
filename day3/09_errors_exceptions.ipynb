## 1 Syntax Errors and Exceptions

**Syntax errors** occur when the code does not conform to Python's grammar rules (e.g. missing parenthesis, wrong indentation). The interpreter shows the line and position; you must correct the code. **Exceptions** occur at **runtime**: e.g. **ZeroDivisionError** (division by 0), **TypeError** (wrong type), **ValueError** (invalid value, e.g. int("abc")), **FileNotFoundError** (file not found), **KeyError** (key missing in dictionary), **IndexError** (index out of list range). With **try/except** you can react to exceptions and let the program continue in a controlled manner.


```python
# Example: ValueError and ZeroDivisionError
try:
    x = int("42")
    print(x + 1)
except ValueError:
    print("Could not convert to int")

try:
    print(10 / 0)
except ZeroDivisionError:
    print("Division by zero is not allowed.")
```

### 1.1 Multiple except Blocks

You can catch different exception types separately: **except ValueError:**, **except TypeError:** etc. The blocks are checked from top to bottom; the first matching one is executed. **except Exception:** catches (almost) all exceptions – useful for logging, but you lose the information about the exact type. Better: specific types first, general one at the end.


```python
# Usage: Different exceptions
def sicher_int(s):
    try:
        return int(s)
    except ValueError:
        return None

print(sicher_int("123"))
print(sicher_int("abc"))
```

## 2 try, except, else, finally

**try**: Block in which an exception can occur. **except** type: catches the specified exception. **else**: Executed only when **no** exception occurred in the try block – ideal for code that should run only on success. **finally**: Executed **always**, with or without exception – typical for releasing resources (closing file, ending connection). The order is try – except – else – finally.


```python
# Example: else and finally
try:
    z = 10 / 2
except ZeroDivisionError:
    print("Division by zero")
else:
    print("Result:", z)
finally:
    print("Done.")
print("---")
try:
    z = 10 / 0
except ZeroDivisionError:
    print("Division by zero")
else:
    print("Result:", z)
finally:
    print("Done.")
```
