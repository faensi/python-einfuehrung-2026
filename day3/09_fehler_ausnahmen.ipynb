## 1 Syntaxfehler und Ausnahmen

**Syntaxfehler** treten auf, wenn der Code nicht den Grammatikregeln von Python entspricht (z.B. fehlende Klammer, falsche Einrueckung). Der Interpreter zeigt die Zeile und Position an; du musst den Code korrigieren. **Ausnahmen** (Exceptions) treten zur **Laufzeit** auf: z.B. **ZeroDivisionError** (Division durch 0), **TypeError** (falscher Typ), **ValueError** (ungueltiger Wert, z.B. int("abc")), **FileNotFoundError** (Datei nicht gefunden), **KeyError** (Schluessel in Dictionary fehlt), **IndexError** (Index ausserhalb der Liste). Mit **try/except** kannst du auf Ausnahmen reagieren und das Programm kontrolliert weiterlaufen lassen.


```python
# Beispiel: ValueError und ZeroDivisionError
try:
    x = int("42")
    print(x + 1)
except ValueError:
    print("Konnte nicht in int umwandeln")

try:
    print(10 / 0)
except ZeroDivisionError:
    print("Division durch Null ist nicht erlaubt.")
```

### 1.1 Mehrere except-Bloecke

Du kannst verschiedene Ausnahmetypen getrennt abfangen: **except ValueError:**, **except TypeError:** usw. Die Bloecke werden von oben nach unten geprueft; der erste passende wird ausgefuehrt. **except Exception:** faengt (fast) alle Ausnahmen – nuetzlich zum Loggen, aber du verlierst die Information ueber den genauen Typ. Besser: konkrete Typen zuerst, allgemeine am Ende.


```python
# Verwendung: Verschiedene Ausnahmen
def sicher_int(s):
    try:
        return int(s)
    except ValueError:
        return None

print(sicher_int("123"))
print(sicher_int("abc"))
```

## 2 try, except, else, finally

**try**: Block, in dem eine Ausnahme auftreten kann. **except** Typ: faengt die genannte Ausnahme. **else**: Wird nur ausgefuehrt, wenn **keine** Ausnahme im try-Block aufgetreten ist – ideal fuer Code, der nur bei Erfolg laufen soll. **finally**: Wird **immer** ausgefuehrt, ob mit oder ohne Ausnahme – typisch zum Freigeben von Ressourcen (Datei schliessen, Verbindung beenden). Die Reihenfolge ist try – except – else – finally.


```python
# Beispiel: else und finally
try:
    z = 10 / 2
except ZeroDivisionError:
    print("Division durch Null")
else:
    print("Ergebnis:", z)
finally:
    print("Fertig.")
print("---")
try:
    z = 10 / 0
except ZeroDivisionError:
    print("Division durch Null")
else:
    print("Ergebnis:", z)
finally:
    print("Fertig.")
```

### 2.1 Dateizugriff und FileNotFoundError

Beim Oeffnen einer Datei mit **open(pfad)** kann **FileNotFoundError** auftreten, wenn die Datei nicht existiert. Statt das Programm abstuerzen zu lassen, kannst du try/except nutzen und eine verstaendliche Meldung ausgeben oder einen Standardwert verwenden.


```python
# Verwendung: Datei oeffnen mit Fehlerbehandlung
pfad = "data/beispiel_text.txt"
try:
    with open(pfad, "r") as f:
        inhalt = f.read()
    print("Erste 50 Zeichen:", inhalt[:50])
except FileNotFoundError:
    print("Datei nicht gefunden:", pfad)
```
