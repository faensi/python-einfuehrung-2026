## 1 NumPy

**NumPy** provides multidimensional **arrays** and fast numerical operations. It is the foundation of many scientific libraries. You create an array with **np.array(list)**. Arithmetic (+, -, *, /) works element-wise. Many functions (mean, sum, min, max) are available as methods or as np.function. **np.arange** and **np.linspace** generate number ranges. NumPy arrays have a **shape** and a **dtype** (data type).


```python
# Example: NumPy arrays
import numpy as np

a = np.array([1, 2, 3, 4, 5])
print(a)
print(a.mean(), a.sum(), a.max(), a.min())
print(a * 2)

m = np.array([[1, 2], [3, 4]])
print(m)
print(m.shape)
```

### 1.1 arange and linspace

**np.arange(start, stop, step)** generates values from start up to but not including stop with step (similar to range). **np.linspace(start, stop, num)** generates a fixed number of evenly spaced values between start and stop (endpoints inclusive). Very useful for x-axes in plots.


```python
# Usage: arange and linspace
print(np.arange(0, 10, 2))
print(np.linspace(0, 1, 5))
x = np.linspace(0, 2 * np.pi, 50)
print(x.shape)
```

## 2 Matplotlib - Data Visualization

**Matplotlib** is the most widespread library for graphics. **import matplotlib.pyplot as plt**. **plt.plot(x, y)** draws a line; **plt.xlabel**, **plt.ylabel**, **plt.title** label the axes and the plot. **plt.grid(True)** enables a grid. **plt.show()** displays the figure (in Jupyter often automatic). You can draw multiple curves in one plot (multiple plt.plot calls) or read data from files (e.g. data/measurements_data.csv) and visualize.


```python
# Example: Simple plot
import matplotlib.pyplot as plt

x = np.linspace(0, 2 * np.pi, 50)
y = np.sin(x)
plt.plot(x, y)
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.title("Sine")
plt.grid(True)
plt.show()
```

### 2.1 Plotting data from data/measurements_data.csv

The file **data/measurements_data.csv** contains columns: Timestamp, Temperature, Humidity, Pressure. You can read it with the **csv** module, put the temperature values into a list and display with **plt.plot(indices, temperatures)** (indices = 0, 1, 2, ... or hour). This shows the typical workflow: read data, convert to arrays/lists, visualize with Matplotlib.


```python
# Usage: Read data/measurements_data.csv and plot temperature
import csv

stunden = []
temperatures = []
with open("data/measurements_data.csv", "r", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    for row in reader:
        stunden.append(len(stunden))
        temperatures.append(float(row["Temperature"]))

plt.figure()
plt.plot(stunden, temperatures)
plt.xlabel("Hour")
plt.ylabel("Temperature (C)")
plt.title("Temperature over time (data/measurements_data.csv)")
plt.grid(True)
plt.show()
```

## 3 SciPy - Brief Overview

**SciPy** builds on NumPy and provides algorithms for optimization, integration, linear algebra, **statistics** (scipy.stats), signal processing, interpolation etc. Example: **scipy.stats** contains mean, standard deviation, distributions. **scipy.interpolate** can estimate intermediate values. For getting started it is enough to know that SciPy is for scientific computations; details you learn in advanced courses.


```python
# Usage: scipy.stats for mean and standard deviation
from scipy import stats

daten = [23.5, 24.1, 23.8, 24.3, 25.0, 24.5, 23.2]
print("Mean:", stats.tmean(daten))
print("Standard deviation:", stats.tstd(daten))
```
