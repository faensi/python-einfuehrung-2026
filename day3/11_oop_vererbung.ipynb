## 1 Vererbung und super()

Eine **Unterklasse** erbt von einer **Basisklasse**, indem du sie in Klammern angibst: **class Kind(Eltern):**. Die Unterklasse hat alle Attribute und Methoden der Elternklasse (sofern nicht ueberschrieben). Mit **super()** rufst du Methoden der Elternklasse auf – besonders wichtig im Konstruktor: **super().__init__(argumente)**, damit die Basisklasse initialisiert wird. So vermeidest du doppelten Code und erweiterst bestehende Funktionalitaet.


```python
# Beispiel: Vererbung
class Tier:
    def __init__(self, name):
        self.name = name

    def laut(self):
        return "???"

class Hund(Tier):
    def __init__(self, name, rasse):
        super().__init__(name)
        self.rasse = rasse

    def laut(self):
        return "Wau!"

h = Hund("Bello", "Labrador")
print(h.name, h.rasse, h.laut())
```

### 1.1 Methoden ueberschreiben

Wenn du in der Unterklasse eine Methode mit demselben Namen definierst, **ueberschreibst** du die Elternmethode. Du kannst in der Unterklasse super().methode() aufrufen, um die Eltern-Implementierung zu nutzen und zusaetzlich eigenen Code auszufuehren (z.B. zusaetzliche Ausgabe vor oder nach dem Aufruf).


```python
# Verwendung: super() in eigener Methode
class Tier:
    def __init__(self, name):
        self.name = name

    def info(self):
        return self.name

class Katze(Tier):
    def __init__(self, name, farbe):
        super().__init__(name)
        self.farbe = farbe

    def info(self):
        return super().info() + ", Farbe: " + self.farbe

k = Katze("Mimi", "schwarz")
print(k.info())
```

## 2 Mehrfachvererbung

Eine Klasse kann von **mehreren** Basisklassen erben: **class C(A, B):**. Python bestimmt die **Method Resolution Order (MRO)**: In welcher Reihenfolge werden Basisklassen durchsucht, wenn eine Methode aufgerufen wird? Du siehst sie mit **Klassenname.__mro__**. Bei Mehrfachvererbung kann es zu Konflikten kommen (z.B. wenn A und B dieselbe Methode haben); super() folgt der MRO. Fuer den Einstieg reicht oft einfache Vererbung (eine Basisklasse).


```python
# Beispiel: Mehrfachvererbung und MRO
class A:
    def meth(self):
        return "A"

class B:
    def meth(self):
        return "B"

class C(A, B):
    pass

c = C()
print(c.meth())
print(C.__mro__)
```

```python
# Musterloesung (a) und (b)
class Tier:
    def __init__(self, name):
        self.name = name

    def laut(self):
        return "???"

class Katze(Tier):
    def __init__(self, name, farbe):
        super().__init__(name)
        self.farbe = farbe

    def laut(self):
        return "Miau!"

k = Katze("Mimi", "schwarz")
print(k.name, k.farbe, k.laut())

# Musterloesung (c)
class Figur:
    def __init__(self, name):
        self.name = name

    def beschreibung(self):
        return self.name

class Rechteck(Figur):
    def __init__(self, name, breite, hoehe):
        super().__init__(name)
        self.breite = breite
        self.hoehe = hoehe

    def beschreibung(self):
        return "{}: {} x {}".format(self.name, self.breite, self.hoehe)

r = Rechteck("R1", 4, 5)
print(r.beschreibung())

# Musterloesung (d)
class Person:
    def __init__(self, name, alter):
        self.name = name
        self.alter = alter

class Angestellter(Person):
    def __init__(self, name, alter, gehalt):
        super().__init__(name, alter)
        self.gehalt = gehalt

    def info(self):
        return "{}, {}, Gehalt: {}".format(self.name, self.alter, self.gehalt)

a = Angestellter("Anna", 30, 50000)
print(a.info())
```
