## 1 Klassen und Instanzen

In Python ist (fast) alles ein Objekt. Mit **class Name:** definierst du einen eigenen Typ. Eine **Instanz** erzeugst du durch Aufruf der Klasse: **obj = Name()**. Die spezielle Methode **__init__(self, ...)** heisst **Konstruktor**: Sie wird beim Erzeugen einer Instanz aufgerufen. Der erste Parameter **self** verweist auf die Instanz selbst; mit **self.attribut = wert** legst du **Instanzattribute** an, die zu diesem konkreten Objekt gehoeren.


```python
# Beispiel: Einfache Klasse
class Person:
    def __init__(self, name, alter):
        self.name = name
        self.alter = alter

    def beschreibung(self):
        return "{} ist {} Jahre alt.".format(self.name, self.alter)

p = Person("Anna", 30)
print(p.name)
print(p.beschreibung())
```

### 1.1 Methoden und self

Methoden sind Funktionen, die zur Klasse gehoeren. Die erste Parameter ist immer **self**. Beim Aufruf **obj.methode()** wird self automatisch mit obj gefuellt. Ueber self greifst du auf andere Attribute und Methoden der Instanz zu.


```python
# Verwendung: Weitere Methode
class Probe:
    def __init__(self, name, wert):
        self.name = name
        self.wert = wert

    def ist_im_messbereich(self, min_val=0, max_val=100):
        return min_val <= self.wert <= max_val

p = Probe("Probe_A", 17.5)
print(p.ist_im_messbereich(0, 50))
p2 = Probe("Probe_B", 25.0)
print(p2.ist_im_messbereich(20, 30))
```

## 2 Klassenattribute und Properties

**Klassenattribute** werden direkt unter der class-Zeile definiert (z.B. zaehler = 0). Sie gehoeren zur Klasse und werden von allen Instanzen geteilt. Aenderst du sie ueber **Klassenname.attribut**, aendert sich der Wert fuer alle. **Instanzattribute** (self.x) gehoeren nur zur jeweiligen Instanz. Mit **@property** definierst du eine Methode, die wie ein Attribut gelesen wird (ohne Klammern); nuetzlich fuer berechnete Werte oder Validierung.


```python
# Beispiel: Klassenattribut und property
class Kreis:
    pi = 3.14159

    def __init__(self, radius):
        self.radius = radius

    @property
    def flaeche(self):
        return self.pi * self.radius ** 2

k = Kreis(5)
print(Kreis.pi)
print(k.flaeche)
print(k.radius)
```

### 2.1 Wann Klassen- vs. Instanzattribut?

Klassenattribute eignen sich fuer Werte, die fuer **alle** Instanzen gleich sind (z.B. Konstanten, Zaehler). Instanzattribute fuer Daten, die pro Objekt unterschiedlich sind (Name, Alter, Radius). Ein Klassenattribut kann als Default dienen, den du in __init__ mit self.attribut = wert ueberschreiben kannst.


```python
# Verwendung: Zaehler als Klassenattribut
class Rechteck:
    anzahl = 0

    def __init__(self, breite, hoehe):
        self.breite = breite
        self.hoehe = hoehe
        Rechteck.anzahl += 1

    def flaeche(self):
        return self.breite * self.hoehe

r1 = Rechteck(3, 4)
r2 = Rechteck(2, 5)
print(Rechteck.anzahl)
print(r1.flaeche(), r2.flaeche())
```

```python
# Musterloesung (a) und (b)
class Rechteck:
    anzahl = 0

    def __init__(self, breite, hoehe):
        self.breite = breite
        self.hoehe = hoehe
        Rechteck.anzahl += 1

    def flaeche(self):
        return self.breite * self.hoehe

r1 = Rechteck(4, 5)
r2 = Rechteck(2, 6)
print(r1.flaeche())
print(Rechteck.anzahl)

# Musterloesung (c)
class Buch:
    def __init__(self, titel, seiten):
        self.titel = titel
        self.seiten = seiten

    @property
    def seiten_pro_seite_hundert(self):
        return 100.0 / self.seiten

b = Buch("Python", 250)
print(b.seiten_pro_seite_hundert)

# Musterloesung (d)
class Messwert:
    def __init__(self, wert, einheit="C"):
        self.wert = wert
        self.einheit = einheit

    def beschreibung(self):
        return "{:.2f} {}".format(self.wert, self.einheit)

m1 = Messwert(23.5)
m2 = Messwert(18.2, "F")
print(m1.beschreibung(), m2.beschreibung())
```
