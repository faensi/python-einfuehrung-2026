## 1 Classes and Instances

In Python (almost) everything is an object. With **class Name:** you define your own type. You create an **instance** by calling the class: **obj = Name()**. The special method **__init__(self, ...)** is the **constructor**: it is called when creating an instance. The first parameter **self** refers to the instance itself; with **self.attribute = value** you create **instance attributes** that belong to this specific object.


```python
# Example: Simple class
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def beschreibung(self):
        return "{} is {} years old.".format(self.name, self.age)

p = Person("Anna", 30)
print(p.name)
print(p.beschreibung())
```

## 2 Class Attributes and Properties

**Class attributes** are defined directly under the class line (e.g. counter = 0). They belong to the class and are shared by all instances. If you change them via **ClassName.attribute**, the value changes for all. **Instance attributes** (self.x) belong only to the respective instance. With **@property** you define a method that is read like an attribute (without parentheses); useful for computed values or validation.


```python
# Example: Class attribute and property
class Circle:
    pi = 3.14159

    def __init__(self, radius):
        self.radius = radius

    @property
    def area(self):
        return self.pi * self.radius ** 2

k = Circle(5)
print(Kreis.pi)
print(k.area)
print(k.radius)
```

### 2.1 When Class vs. Instance Attribute?

Class attributes are suitable for values that are the same for **all** instances (e.g. constants, counters). Instance attributes for data that differs per object (name, age, radius). A class attribute can serve as default, which you can override in __init__ with self.attribute = value.


```python
# Usage: Counter as class attribute
class Rectangle:
    count = 0

    def __init__(self, width, hoehe):
        self.width = width
        self.hoehe = hoehe
        Rectangle.count += 1

    def area(self):
        return self.width * self.hoehe

r1 = Rectangle(3, 4)
r2 = Rectangle(2, 5)
print(Rectangle.count)
print(r1.area(), r2.area())
```
