## 1 Inheritance and super()

A **subclass** inherits from a **base class** by specifying it in parentheses: **class Child(Parent):**. The subclass has all attributes and methods of the parent class (unless overridden). With **super()** you call methods of the parent class – especially important in the constructor: **super().__init__(arguments)**, so the base class is initialized. This avoids code duplication and extends existing functionality.


```python
# Example: Inheritance
class Animal:
    def __init__(self, name):
        self.name = name

    def make_sound(self):
        return "???"

class Hund(Animal):
    def __init__(self, name, rasse):
        super().__init__(name)
        self.rasse = rasse

    def make_sound(self):
        return "Wau!"

h = Dog("Bello", "Labrador")
print(h.name, h.breed, h.make_sound())
```

### 1.1 Overriding Methods

When you define a method with the same name in the subclass, you **override** the parent method. You can call super().method() in the subclass to use the parent implementation and additionally execute your own code (e.g. additional output before or after the call).


```python
# Usage: super() in own method
class Animal:
    def __init__(self, name):
        self.name = name

    def info(self):
        return self.name

class Cat(Animal):
    def __init__(self, name, color):
        super().__init__(name)
        self.color = color

    def info(self):
        return super().info() + ", Color: " + self.color

k = Cat("Mimi", "schwarz")
print(k.info())
```

## 2 Multiple Inheritance

A class can inherit from **multiple** base classes: **class C(A, B):**. Python determines the **Method Resolution Order (MRO)**: In what order are base classes searched when a method is called? You can see it with **ClassName.__mro__**. With multiple inheritance conflicts can occur (e.g. when A and B have the same method); super() follows the MRO. For getting started, simple inheritance (one base class) is often sufficient.


```python
# Example: Multiple inheritance and MRO
class A:
    def meth(self):
        return "A"

class B:
    def meth(self):
        return "B"

class C(A, B):
    pass

c = C()
print(c.meth())
print(C.__mro__)
```
