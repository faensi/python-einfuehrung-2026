## 1 NumPy

**NumPy** bietet mehrdimensionale **Arrays** und schnelle numerische Operationen. Es ist die Grundlage vieler wissenschaftlicher Bibliotheken. Ein Array erzeugst du mit **np.array(liste)**. Arithmetik (+, -, *, /) wirkt elementweise. Viele Funktionen (mean, sum, min, max) sind als Methoden oder als np.funktion verfuegbar. **np.arange** und **np.linspace** erzeugen Zahlenbereiche. NumPy-Arrays haben einen **shape** (Form) und einen **dtype** (Datentyp).


```python
# Beispiel: NumPy-Arrays
import numpy as np

a = np.array([1, 2, 3, 4, 5])
print(a)
print(a.mean(), a.sum(), a.max(), a.min())
print(a * 2)

m = np.array([[1, 2], [3, 4]])
print(m)
print(m.shape)
```

### 1.1 arange und linspace

**np.arange(start, stop, step)** erzeugt Werte von start bis vor stop mit Schritt step (aehnlich range). **np.linspace(start, stop, anzahl)** erzeugt eine feste Anzahl gleichmaessig verteilter Werte zwischen start und stop (Endpunkte inklusive). Sehr nuetzlich fuer x-Achsen in Plots.


```python
# Verwendung: arange und linspace
print(np.arange(0, 10, 2))
print(np.linspace(0, 1, 5))
x = np.linspace(0, 2 * np.pi, 50)
print(x.shape)
```

## 2 Matplotlib - Datenvisualisierung

**Matplotlib** ist die verbreitetste Bibliothek fuer Grafiken. **import matplotlib.pyplot as plt**. **plt.plot(x, y)** zeichnet eine Linie; **plt.xlabel**, **plt.ylabel**, **plt.title** beschriften die Achsen und den Plot. **plt.grid(True)** schaltet ein Gitter ein. **plt.show()** zeigt die Figur an (in Jupyter oft automatisch). Du kannst mehrere Kurven in einem Plot zeichnen (mehrere plt.plot-Aufrufe) oder Daten aus Dateien (z.B. data/measurements_data.csv) einlesen und visualisieren.


```python
# Beispiel: Einfacher Plot
import matplotlib.pyplot as plt

x = np.linspace(0, 2 * np.pi, 50)
y = np.sin(x)
plt.plot(x, y)
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.title("Sinus")
plt.grid(True)
plt.show()
```

### 2.1 Daten aus data/measurements_data.csv plotten

Die Datei **data/measurements_data.csv** enthaelt Spalten: Timestamp, Temperature, Humidity, Pressure. Du kannst sie mit dem **csv**-Modul einlesen, die Temperaturwerte in eine Liste packen und mit **plt.plot(indices, temperaturen)** darstellen (indices = 0, 1, 2, ... oder Stunde). So siehst du den typischen Ablauf: Daten einlesen, in Arrays/Listen umwandeln, mit Matplotlib visualisieren.


```python
# Verwendung: data/measurements_data.csv einlesen und Temperatur plotten
import csv

stunden = []
temperaturen = []
with open("data/measurements_data.csv", "r", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    for row in reader:
        stunden.append(len(stunden))
        temperaturen.append(float(row["Temperature"]))

plt.figure()
plt.plot(stunden, temperaturen)
plt.xlabel("Stunde")
plt.ylabel("Temperatur (C)")
plt.title("Temperaturverlauf (data/measurements_data.csv)")
plt.grid(True)
plt.show()
```

## 3 SciPy - Kurzer Einblick

**SciPy** baut auf NumPy auf und bietet Algorithmen fuer Optimierung, Integration, lineare Algebra, **Statistik** (scipy.stats), Signalverarbeitung, Interpolation usw. Beispiel: **scipy.stats** enthaelt Mittelwert, Standardabweichung, Verteilungen. **scipy.interpolate** kann Zwischenwerte schaetzen. Fuer den Einstieg reicht es, zu wissen, dass SciPy fuer wissenschaftliche Berechnungen da ist; Einzelheiten lernst du in fortgeschrittenen Kursen.


```python
# Verwendung: scipy.stats fuer Mittelwert und Standardabweichung
from scipy import stats

daten = [23.5, 24.1, 23.8, 24.3, 25.0, 24.5, 23.2]
print("Mittelwert:", stats.tmean(daten))
print("Standardabweichung:", stats.tstd(daten))
```
