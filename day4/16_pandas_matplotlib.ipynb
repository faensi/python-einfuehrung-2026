## 1 Tabular Data: Why Pandas?

**Tabular data** has rows and columns (like a spreadsheet). Common formats are **CSV** (comma-separated) and **JSON** (structured text). Pandas provides **DataFrames**—tables with column names, row indices, and built-in methods for filtering, grouping, and aggregating. It is the standard tool for data analysis in Python.


## 2 Reading CSV Files

**pd.read_csv(filename)** loads a CSV into a DataFrame. Columns become named fields; you can select them with **df['col']** or **df.col**. Use **df.head()** to peek at the first rows, **df.info()** for columns and types, **df.describe()** for numeric summaries.


```python
import pandas as pd

# Load CSV (150 rows of sensor data)
df = pd.read_csv("data/sensor_readings.csv")
print(df.head(10))
print("\nShape:", df.shape)
print("\nInfo:")
df.info()
```

```python
# Basic statistics for numeric columns
print(df.describe())

# Select a column
temps = df["temperature"]
print("\nMean temperature:", temps.mean())
```

## 3 Reading JSON Files

**pd.read_json(filename)** loads JSON. If the JSON is a list of objects (records), each object becomes a row and keys become columns. JSON is common for web APIs and configuration; Pandas can convert it to a DataFrame for analysis.


```python
# Load JSON (150 product entries)
products = pd.read_json("data/products.json")
print(products.head(10))
print("\nShape:", products.shape)
products.info()
```

```python
# Group by category and aggregate
by_cat = products.groupby("category").agg({
    "price": ["mean", "min", "max"],
    "stock": "sum"
})
print(by_cat)
```

## 4 Plotting with Matplotlib

Combine Pandas with **matplotlib.pyplot** for visualization. Use **plt.plot(df['x'], df['y'])** or **df['col'].plot()**. Common types: **plt.plot** (line), **plt.bar** (bars), **plt.scatter** (points). Label axes and add a title for clarity.


```python
import matplotlib.pyplot as plt

# Line plot: temperature over time (first 50 rows)
sub = df.head(50)
plt.figure(figsize=(10, 4))
plt.plot(sub["id"], sub["temperature"], label="Temperature")
plt.xlabel("Reading index")
plt.ylabel("Temperature (°C)")
plt.title("Temperature over time (sensor_readings.csv)")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Bar chart: total stock per category
stock_by_cat = products.groupby("category")["stock"].sum()
plt.figure(figsize=(8, 4))
plt.bar(stock_by_cat.index, stock_by_cat.values)
plt.xlabel("Category")
plt.ylabel("Total stock")
plt.title("Stock by category (products.json)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Scatter: price vs stock
plt.figure(figsize=(6, 5))
plt.scatter(products["price"], products["stock"], alpha=0.6)
plt.xlabel("Price")
plt.ylabel("Stock")
plt.title("Price vs stock")
plt.show()
```

## 5 CSV vs JSON

| Format | Typical use | Pandas function |
|--------|-------------|-----------------|
| CSV | Spreadsheets, exports, simple tables | pd.read_csv() |
| JSON | APIs, nested structures, config | pd.read_json() |

Both can represent tabular data; CSV is simpler for flat tables, JSON for hierarchical or API responses.
