## 1 Package Managers and Distributions

### Conda
**Conda** is a cross-platform package and environment manager. It installs packages (including non-Python ones like C libraries) and resolves dependencies. Often used for scientific Python (NumPy, SciPy, etc.) because it handles binary dependencies well.

### Anaconda
**Anaconda** is a full distribution: conda + Python + 600+ pre-installed packages (data science, Jupyter, etc.) + Anaconda Navigator GUI. Convenient for beginners but **heavy** (~3–5 GB). In reproducible scientific workflows, **Miniconda is often preferred** because it keeps environments small and explicit—you install only what you need, which reduces conflicts and eases sharing.

### Miniconda
**Miniconda** is a minimal installer: conda + Python + base dependencies (~a few hundred MB). You add packages per project. Preferred for **reproducible research**: create project-specific environments, pin versions in `environment.yml`, and share the same setup across machines.

### Mamba
**Mamba** is a faster, drop-in replacement for conda. Same commands (`mamba install`, `mamba create`), written in C++, uses parallel solving. Often used when conda is slow. Works with Miniconda/Anaconda.

### Rattler
**Rattler** is a Rust library (and tool) for the conda ecosystem. Used by tools like **Pixi** and **uv** for faster environment solving. You typically interact with it indirectly.


## 2 Package Sources: conda-forge and Defaults

### conda-forge
**conda-forge** is a community-driven channel of conda packages. Many scientific and open-source packages are maintained there. Add it with `conda config --add channels conda-forge` or `conda install -c conda-forge <package>`. Often preferred for up-to-date and comprehensive packages.

### Defaults (Anaconda)
The **defaults** channel is Anaconda’s official channel. Some packages are only there; others are duplicated on conda-forge. Mixing channels can cause conflicts—many users standardize on conda-forge.


## 3 Python Package Installers (pip, uv)

### pip
**pip** is Python’s default package installer. Uses **PyPI** (Python Package Index). `pip install numpy` installs from PyPI. Works inside conda environments too, but for mixed conda+pip environments, install conda packages first, then pip.

### uv
**uv** is a fast, Rust-based Python package installer and resolver (by Astral). Can replace pip for many use cases. Often used with modern workflows and tools like **Pixi**.


## 4 Virtual Environments

### venv
**venv** is Python’s built-in virtual environment tool. `python -m venv myenv` creates an isolated environment. Use `source myenv/bin/activate` (Unix) or `myenv\Scripts\activate` (Windows). Good for pip-only projects.

### virtualenv
**virtualenv** is a third-party tool similar to venv, with extra features. Sometimes used when venv is not enough.

### Conda environments
**Conda environments** are separate from venv: `conda create -n myenv python=3.11` and `conda activate myenv`. They can include non-Python dependencies. `environment.yml` or `conda env export` capture an environment for reproducibility.


## 5 IDEs and Editors

### PyCharm
**PyCharm** is a full IDE by JetBrains. Features: code completion, refactoring, debugging, integrated terminals. Community Edition is free; Professional adds web frameworks, databases, etc. Popular for larger Python projects.

### VS Code / Cursor
**VS Code** (and **Cursor**) are lightweight editors with Python extensions. Run code, debug, use Jupyter notebooks. Popular for general development and data science.

### Jupyter Notebook / JupyterLab
**Jupyter Notebook** provides interactive cells (code + markdown) in the browser. Great for exploration, teaching, and analysis. **JupyterLab** is the next-generation interface: notebooks, terminals, file browser in one. Both use **IPython** as the kernel.


## 6 Common Terms Cheat Sheet

| Term | Meaning |
|------|--------|
| **conda** | Package and environment manager |
| **pip** | Python package installer (PyPI) |
| **venv** | Built-in Python virtual environment |
| **environment.yml** | Conda environment definition (reproducible) |
| **requirements.txt** | pip package list (often pinned) |
| **kernel** | Execution backend (e.g. IPython for Jupyter) |
| **channel** | Source of conda packages (conda-forge, defaults) |
| **PyPI** | Python Package Index (pip’s default source) |
| **Anaconda** | Full distribution with many pre-installed packages |
| **Miniconda** | Minimal conda + Python install |
| **Mamba** | Faster conda alternative |
