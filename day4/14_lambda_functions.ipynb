## 1 What is lambda?

A **lambda** creates a small anonymous function in one line. Syntax: **lambda** args **:** expression. The expression is evaluated and returned; no return needed. Use lambdas for short, one-off functions (e.g. as an argument to sorted or map). For anything longer, use **def**.


```python
# Example: lambda vs def
def double_def(x):
    return x * 2

double_lambda = lambda x: x * 2

print(double_def(5))
print(double_lambda(5))
print((lambda a, b: a + b)(3, 4))
```

### 1.1 Lambda with sorted()

**sorted(iterable, key=...)** sorts by the result of the key function. A lambda is often used for key. Example: sort a list of tuples by the second element.


```python
# Usage: sorted with key
pairs = [(1, "one"), (3, "three"), (2, "two")]
print(sorted(pairs))
print(sorted(pairs, key=lambda p: p[1]))

words = ["banana", "apple", "cherry"]
print(sorted(words, key=lambda w: len(w)))
```

## 2 filter() and map()

**filter(function, iterable)** keeps only elements where the function returns True. **map(function, iterable)** applies the function to each element. Both return iterators; use **list()** to see the result.


```python
# Example: filter and map with lambda
numbers = [1, 2, 3, 4, 5, 6]
evens = list(filter(lambda n: n % 2 == 0, numbers))
squared = list(map(lambda n: n ** 2, numbers))

print("Evens:", evens)
print("Squared:", squared)
```

### 2.1 When to Use lambda

**Use lambda** for short, single-use functions passed to sorted, filter, map. **Use def** when the logic is longer or should be reused. Lambda can only contain a single expression.


```python
# Usage: Practical example
data = [("Alice", 30), ("Bob", 25), ("Carol", 35)]
by_age = sorted(data, key=lambda x: x[1])
print("Sorted by age:", by_age)

names = ["python", "java", "rust"]
titles = list(map(lambda s: s.title(), names))
print("Titled:", titles)
```
