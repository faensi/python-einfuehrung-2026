## 1 What is an API?

An **API** (Application Programming Interface) lets programs communicate over the network. A **REST API** uses URLs and HTTP methods: **GET** (fetch data), **POST** (send data), **PUT/PATCH** (update), **DELETE** (remove). Responses are often in **JSON** (structured text). You send a request, get a response (e.g. status code 200 = OK, 404 = not found).


## 2 The requests Library

**requests** is the standard library for HTTP in Python. **requests.get(url)** sends a GET request; **response.json()** parses JSON. **response.status_code** gives the HTTP status. Install with **pip install requests** (or conda).


```python
import requests

# Fetch a single post (ID 1)
url = "https://jsonplaceholder.typicode.com/posts/1"
response = requests.get(url)

print("Status code:", response.status_code)
data = response.json()
print("Data:", data)
```

## 3 Fetching a List of Items

Many APIs return a list of objects. You can loop over them or convert to a Pandas DataFrame for analysis.


```python
# Fetch all posts
url = "https://jsonplaceholder.typicode.com/posts"
response = requests.get(url)

if response.status_code == 200:
    posts = response.json()
    print(f"Fetched {len(posts)} posts")
    print("First post:", posts[0])
else:
    print("Error:", response.status_code)
```

```python
# Use Pandas to work with API data
import pandas as pd

df = pd.DataFrame(posts)
print(df.head())
print("\nPosts per user:")
print(df["userId"].value_counts().head())
```

## 4 Filtering with Query Parameters

APIs often support **query parameters** in the URL (e.g. `?userId=1`). These filter or limit the results.


```python
# Get only posts from user 3
url = "https://jsonplaceholder.typicode.com/posts"
params = {"userId": 3}
response = requests.get(url, params=params)

posts_user3 = response.json()
print(f"User 3 has {len(posts_user3)} posts")
print(posts_user3[0]["title"])
```

## 5 Error Handling

Check **response.raise_for_status()** to raise an exception on 4xx/5xx errors. Or check **response.status_code** manually before parsing.


```python
response = requests.get("https://jsonplaceholder.typicode.com/posts/9999")
if response.status_code == 200:
    print(response.json())
else:
    print(f"Request failed with status {response.status_code}")
```
