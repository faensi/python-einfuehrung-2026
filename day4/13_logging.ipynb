## 1 Why Logging?

**print()** is fine for quick debugging, but for real applications **logging** is better: you can control output levels (e.g. show only WARNING and above in production), send messages to files, add timestamps, and keep debug output available without cluttering the console. The **logging** module is part of the standard library.


```python
# Example: Simple logging (uses default configuration)
import logging
logging.basicConfig(level=logging.INFO)

logging.debug("This won't show - DEBUG is below INFO")
logging.info("Application started")
logging.warning("Low disk space")
logging.error("Connection failed")
```

### 1.1 Log Levels

Log levels from lowest to highest: **DEBUG** (details), **INFO** (general info), **WARNING** (something unexpected), **ERROR** (serious problem), **CRITICAL** (program may stop). When you set `level=logging.INFO`, messages at INFO and above (INFO, WARNING, ERROR, CRITICAL) are shown; DEBUG is filtered out.


```python
# Usage: Different levels
import logging

logging.basicConfig(level=logging.INFO, format="%(levelname)s: %(message)s")
logging.debug("Debug message")
logging.info("Info message")
logging.warning("Warning message")
```

## 2 Logging to a File

With **logging.basicConfig(filename="app.log")** messages go to a file instead of the console. You can combine **format** (e.g. `%(asctime)s - %(levelname)s - %(message)s`) for timestamps. Use `filemode="a"` to append instead of overwriting.


```python
# Usage: Log to file
import logging

logging.basicConfig(
    filename="data/app.log",
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)
logging.info("Program started")
logging.warning("Check the log file: data/app.log")
print("Messages written to data/app.log")
```
