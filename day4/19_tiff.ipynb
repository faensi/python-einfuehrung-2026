## 1 What is TIFF?

**TIFF** (Tagged Image File Format, extension .tif or .tiff) is a flexible image format that supports:
- Lossless compression
- Multiple bits per pixel (8, 16, 32-bit)
- Grayscale and multi-channel images
- Multi-page stacks (e.g. microscopy Z-stacks, time series)
- Metadata (tags) for scientific data

It is widely used in microscopy, medical imaging, and geospatial data. In Python, TIFF files are typically read as **NumPy arrays**, so you can apply array operations, statistics, and visualization.


## 2 Loading a TIFF with tifffile

The **tifffile** library reads TIFF files directly into NumPy arrays. **tifffile.imread(path)** returns the image data. Use **tifffile.imwrite(path, array)** to save. For multi-page TIFFs, the array has an extra dimension (e.g. shape `(frames, height, width)`).


```python
import numpy as np
import tifffile
from pathlib import Path

# Create sample TIFF if data/at3_1m4_01.tif does not exist
tiff_path = Path("data/at3_1m4_01.tif")
if not tiff_path.exists():
    tiff_path.parent.mkdir(parents=True, exist_ok=True)
    sample = np.random.randint(0, 255, (64, 64), dtype=np.uint8)
    tifffile.imwrite(str(tiff_path), sample)
    print("Created sample TIFF at", tiff_path)

# Load the TIFF
img = tifffile.imread(str(tiff_path))
print("Shape:", img.shape)
print("dtype:", img.dtype)
print("Min:", img.min(), "Max:", img.max())
```

```python
# Display with matplotlib
import matplotlib.pyplot as plt

plt.figure(figsize=(5, 5))
if img.ndim == 2:
    plt.imshow(img, cmap="gray")
    plt.colorbar()
elif img.ndim == 3:
    plt.imshow(img[0], cmap="gray")  # first frame of stack
    plt.colorbar()
plt.title("at3_1m4_01.tif")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

## 3 Basic Operations on TIFF Data

Because TIFF data is a NumPy array, you can use standard operations: slicing, mean, std, histogram. For a 2D image, `img[y1:y2, x1:x2]` extracts a region. `img.mean()` and `img.std()` give statistics.


```python
# Statistics and region
print("Mean:", img.mean())
print("Std:", img.std())

# Extract center region (example for 2D)
if img.ndim == 2:
    h, w = img.shape
    center = img[h//4 : 3*h//4, w//4 : 3*w//4]
    print("Center region shape:", center.shape)
    print("Center mean:", center.mean())
```
